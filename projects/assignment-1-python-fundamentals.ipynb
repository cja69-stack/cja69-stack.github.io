Problem 1. Calculator: Create a calculator that takes two numbers and an operator (addition, subtraction, multiplication, division) as input and outputs the result. (Bonus: Extend this to handle multiple calculations using a loop).


```python
def calculator(n1 ,n2 ,op):
  if op == "+":
    return n1+n2
  elif op == "-"
    return n1-n2
  elif op == "*"
    return n1*n2
  elif op == "/"
    if n2 == 0:
      print("Cannot divide by 0")
    else:
      return n1/n2
  else:
    print("You didn't enter a valid operation")

result = calculator(1,3,+)
print(result)

```

Problem 3. Create a two-player Rock-Paper-Scissors game that runs in the terminal. The game should handle player input, check for a win condition, and declare a winner or tie.

```python
playing = True
while playing:
  p1 = input("Player 1, what are you choosing, rock, paper, or scissors? ")

  p2 = input("Player 2, What are you choosing, rock, paper, or scissors? ")

  if p1 == p2:
    print("Tie")

  elif p1 == "paper" and p2 == "rock":
    print("Player 1 wins!")

  elif p1 == "rock" and p2 == "scissors":
    print("Player 1 wins!")

  elif p1 == "scissors" and p2 == "paper":
    print("Player 1 wins!")

  elif p2 == "paper" and p1 == "rock":
    print("Player 2 wins!")

  elif p2 == "rock" and p1 == "scissors":
    print("Player 2 wins!")

  elif p2 == "scissors" and p1 == "paper":
    print("Player 2 wins!")

  again = input("Want to play again? ")

  if again != "yes":
    playing = False
    print("Thanks for playing")

```

Problem 4. Create a program to generate the first n digits of the Fibonacci sequence. The program should take input from the user for ‘n’.

```python
fib_length = int(input("How many digits of the fib sequence to generate "))

def fib(n):

  if n == 0:
    return "Invalid Input "

  #edge case 1
  elif n == 1:
    return 0

  #edge case 2
  elif n == 2:
    return [0,1]

  #normal case
  else:
    seq = [0,1]

    for i in range(2,n):
      seq.append(seq[i-1] + seq[1-2])

    return seq

print(fib(fib_length))



```
